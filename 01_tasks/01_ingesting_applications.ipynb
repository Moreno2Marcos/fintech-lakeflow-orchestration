# Notebook Task — Credit Applications Ingestion

**Objective:** Read synthetic fintech credit applications from an existing Delta table and persist the records into a staging Delta table using Databricks and PySpark.

## Why?

**For our project:** Creates the `applications_staging` table from `credit_applications`, providing a simple processing task that Lakeflow Jobs can execute and monitor.

**For Data Engineering:** Demonstrates the separation between data processing logic and workflow orchestration, using reproducible Delta table operations.

#### 01 — Read Source Credit Applications

Read the existing credit applications from Unity Catalog using PySpark. The source table contains six synthetic credit applications.

```python
# Define the source and target Delta tables.
source_table = "workspace.fintech_jobs.credit_applications"
target_table = "workspace.fintech_jobs.applications_staging"

# Read the existing credit applications from Unity Catalog.
applications_df = spark.table(source_table)
```

#### 02 — Write Staging Delta Table

Write the source records to `applications_staging` using Delta Lake. Overwrite mode replaces the existing target data, preventing duplicate accumulation during repeated executions.

```python
# Create or overwrite the target Delta table.
applications_df.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable(target_table)
```

#### 03 — Validate Data Load

Compare the source and target record counts to verify that all six credit applications were loaded into the staging table.

```python
# Confirm that the task processed the expected records.
print(f"Source records: {applications_df.count()}")
print(f"Target records: {spark.table(target_table).count()}")
```

  <h3>💡 Engineering Note — Staging vs. Bronze</h3>

  <p>The <code>applications_staging</code> table stores an intermediate copy of the source data without transformations.</p>

  <p>Although this resembles a Bronze layer, staging tables primarily support intermediate data processing and do not necessarily preserve historical raw data.</p>

  <p><strong>In this project:</strong> Staging prepares a simple table-to-table workload for Lakeflow Jobs orchestration, without implementing a Medallion Architecture.</p>
