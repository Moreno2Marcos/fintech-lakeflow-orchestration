# Fintech Workloads — Failure Recovery and Performance Monitoring

**Objective:** Extend our fintech workflow with a controlled task failure, selective repair, a published dashboard, and operational monitoring.

## Why?

**For our project:** Recover an application transformation without repeating successful ingestion and reporting tasks, then refresh a dashboard after both reporting branches finish.

**For Data Engineering:** Diagnose failed tasks, distinguish repair from a new run, design repeatable writes, inspect query performance, and query job execution history.

#### 01 — Prepare the Recovery Job

Clone the validated dynamic workflow to reuse its task dependencies
and five state reports for the failure recovery and monitoring exercise.

#### 02 — Create the Application Transformation

Add a transformation after all five state reports finish.
Combine their records into an application detail table and a
state summary table for the dashboard.

The application transformation depends on the For each container
and runs after all five state reports succeed. For the controlled
failure test, automatic retries are disabled on this task.

#### 03 — Execute and Diagnose the Controlled Failure

Run the complete recovery job and inspect the failed application
transformation. Identify the missing column while confirming that
the upstream tasks succeeded.

#### 04 — Correct the Code and Repair the Same Run

Correct the customer-name reference in the source notebook.
Repair the failed application transformation within the existing
job run, preserving successful upstream work.

#### 05 — Validate the Repaired Outputs

Compare the repaired application details and state summary
with the source data. Verify the existing installment totals
before creating the dashboard.

#### 06 — Create and Publish the Fintech Dashboard

Create a dashboard using the validated application and installment
tables. Configure its datasets and visualizations, then publish it
for refresh through the recovery job.

#### 07 — Add the Dashboard Refresh Task

Add refreshing_fintech_dashboard to the recovery job.
Configure it to refresh fintech_credit_operations_dashboard
using Serverless Starter Warehouse.

The task depends on transforming_applications_by_state and
transforming_customers_installments, with Run if set to All succeeded.
This ensures both data branches finish successfully before the
dashboard refreshes.
